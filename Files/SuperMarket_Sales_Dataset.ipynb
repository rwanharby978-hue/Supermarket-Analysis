```python
# Data Cleaning Phase
df.columns
```

```python
# Rename Columns
df = df.rename (columns = {'   Customer type' : 'Customer type', '   cogs' : 'COGS','Payment    ':'Payment'} )
```

```python
df.columns = df.columns.str.replace(' ','_')
df.columns
```

```python
# Change data type 
df['Invoice_ID'] = df['Invoice_ID'].str.replace('-', '')
df['Invoice_ID'] = df['Invoice_ID'].astype('int64')
```

```python
df['Invoice_ID'].dtype
```

```python
df['Sales'] = df['Sales'].str.replace('EGP', '')
df['Sales'] = df['Sales'].astype('float64')
```

```python
df['Date'] = pd.to_datetime(df['Date'], errors='coerce')
```

```python
# Making sure all columns have the correct data types
df.dtypes
```

```python
# Handling Missing Values
df['Unit_price'] = df['Unit_price'].fillna(((df['Sales'] - df['Tax_5%']) / df['Quantity']))
```

```python
df['Sales'] = df['Sales'].fillna((df['Unit_price'] * df['Quantity']) + df['Tax_5%'])
```

```python
df['Date'] = df['Date'].fillna("2020-01-01")
```

```python
# Check Duplicates
df.duplicated().sum()
```

```python
df = df.drop_duplicates()
```

```python
# Remove Unnecessary columns
df = df.drop (
    columns=[
        "Gender",
        "Unit_price",
        "Quantity",
        "Tax_5%",
        "Time",
        "gross_margin_percentage",
        "COGS"
    ]
)
```

```python
df['Branch'].value_counts()
```

```python
df['Branch'] = df['Branch'].str.strip().replace({'giza':'Giza','cairo':'Cairo'})
```

```python
df['City'].value_counts()
```

```python
df['Customer_type'].value_counts()
```

```python
df['Product_line'].value_counts()
```

```python
df['Payment'].value_counts()
```

```python
df['Payment'] = df['Payment'].replace({'cash':'Cash','credit card':'Credit card', 'E-wallet':'Ewallet'}) 
```

```python
# Sort data by date
df = df.sort_values('Date')
```

```python
df['Year'] = df['Date'].dt.year
df['Month'] = df['Date'].dt.month
df['Day'] = df['Date'].dt.day
```

```python
# Creating Satisfiction Column
def check_satisfiction(rating):
    if rating >= 7:
      return 'Satisfied'
    else: 
       return 'Not Satisfied'
```

```python
df['Satisfied'] = df['Rating'].apply(check_satisfiction)
```

```python
df.to_csv("SuperMarket_Cleaned.csv", index=False)
```
